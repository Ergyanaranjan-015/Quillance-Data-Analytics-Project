```python
import pandas as pd

# Reference Date
reference_date = pd.Timestamp("2018-01-01")

# RFM Calculation
rfm = df.groupby("Customer ID").agg({
    "Order Date": lambda x: (reference_date - x.max()).days,
    "Order ID": "nunique",
    "Sales": "sum"
})

# Rename Columns
rfm.rename(columns={
    "Order Date": "Recency",
    "Order ID": "Frequency",
    "Sales": "Monetary"
}, inplace=True)

# Round Monetary
rfm["Monetary"] = rfm["Monetary"].round(2)

# Display Result
print("RFM Shape:", rfm.shape)
display(rfm.head(10))
```

```python
# RFM Scoring using Quantiles

rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

display(rfm.head(10))
```

```python
# Combine RFM Scores
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str) +
    rfm["F_Score"].astype(str) +
    rfm["M_Score"].astype(str)
)

# Customer Segmentation
def customer_segment(row):
    if row["R_Score"] >= 4 and row["F_Score"] >= 4 and row["M_Score"] >= 4:
        return "Champions"
    elif row["R_Score"] >= 3 and row["F_Score"] >= 3:
        return "Loyal Customers"
    elif row["R_Score"] >= 4 and row["F_Score"] <= 2:
        return "Recent Customers"
    elif row["R_Score"] <= 2 and row["F_Score"] >= 3:
        return "At Risk"
    elif row["R_Score"] <= 2 and row["F_Score"] <= 2:
        return "Inactive Customers"
    else:
        return "Potential Customers"

rfm["Customer_Segment"] = rfm.apply(customer_segment, axis=1)

display(rfm.head(15))
```

```python
# Add Customer ID as a column
rfm_summary = rfm.reset_index()

# Segment-wise Analysis
segment_analysis = rfm_summary.groupby("Customer_Segment").agg(
    Total_Customers=("Customer ID", "count"),
    Total_Sales=("Monetary", "sum"),
    Average_Recency=("Recency", "mean"),
    Average_Frequency=("Frequency", "mean"),
    Average_Monetary=("Monetary", "mean")
).round(2)

# Sort by Sales
segment_analysis = segment_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

display(segment_analysis)
```

```python
# Save RFM Customer Data
rfm_summary.to_csv(
    "Superstore_RFM_Analysis.csv",
    index=False
)

# Save Segment Summary
segment_analysis.to_csv(
    "Superstore_RFM_Segment_Summary.csv"
)

print("RFM Analysis files saved successfully!")
```
